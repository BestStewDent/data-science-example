# Demo Data Science Exploration

This notebook is a simple narrative walkthrough showing how a data science presentation might explore a synthetic customer dataset.


```python
from pathlib import Path
import pandas as pd

data_dir = Path('../data/sample')
customers = pd.read_csv(data_dir / 'customers.csv', parse_dates=['signup_date'])
orders = pd.read_csv(data_dir / 'orders.csv', parse_dates=['order_date'])
df = customers.merge(orders, on='customer_id', how='left')
df.head()
```

## Key observations

- High-value customers drive a large share of revenue.
- Repeat purchases are a useful signal for retention.
- Regional differences help tell a simple story about market behavior.
